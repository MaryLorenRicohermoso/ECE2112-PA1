## A. WORD ROTATION PROBLEM

```python
def rotate_word(text):  #defines the name of the function and variable
    return text[1: ] + text [0]     #moves the first character to the end
 

print (rotate_word("python"))     #print the outpur using "python"
```

## B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):  #defines the name of the function and variable
   first_name = first_name.lower()      
                                            #changes the letters to lowercase
   last_name = last_name.lower()

   first_name = first_name.replace(" ", "")
                                                #remove space
   last_name = last_name.replace (" ", "")

   return first_name + "." + last_name          #joins the names with a period
              
print(make_username("Ana Maria", "De Leon"))      #print the result
```

## C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):  #defines the name of the function and variable
    first, *middle, last = items       #separates the first, middle, and last elements
    return [last] + middle + [first]

print (swap_bookends([1, 2, 3, 4, 5, 6]))           #print the resullt
```
